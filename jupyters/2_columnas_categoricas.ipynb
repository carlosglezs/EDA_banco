# Gestion de nulos de categoricas

```python
#Generamos un dataset con las columnas categoricas
df_cat = df_limpio[df_limpio.select_dtypes(include="O").columns]
```

Vemos que no tenemos ninguna columna con nulos por encima del 50%. Vamos a gestionar los nulos rellenando esos valores.

```python
#Observamos la dispersión de las variables categoricas
sp.analisis_general_cat(df_cat)
```

Rellenaremos 'Loan', 'Default' y 'y' con la moda. Pues en este caso son muy superiores.
El resto las rellenamos con unknown

```python
# Reemplazar NaN por "no" en las columnas 'loan' y 'default'
df_limpio[['loan', 'default']] = df_limpio[['loan', 'default']].fillna('no')
```

```python
#Ahora podemos reemplazar el resto por "unknown"
categorical_cols = df_limpio.select_dtypes(include=['O']).columns
df_limpio[categorical_cols] = df_limpio[categorical_cols].fillna('unknown')
```

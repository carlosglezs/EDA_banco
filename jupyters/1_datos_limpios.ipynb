```python
#Instalaciones básicas
import pandas as pd
import numpy as np
import datetime
import sys
sys.path.append("..")
from src import soporte as sp

```

```python
#Leer el archivo

pd.set_option('display.max_columns', None)
df_raw = pd.read_csv("../data/bank-additional.csv")
```

```python
#Generamos copia de seguridad
df = df_raw.copy()
```

```python
sp.eda_preliminar(df)
```

```python
#Borramos columna Unnamed:0
df.drop(columns="Unnamed: 0", inplace=True)
```

```python
#Porcentaje de nulos redondeado a 2 de decimales
round(df.isnull().sum()/df.shape[0]*100,2)
```

```python
#Borramos posibles duplicados
df.duplicated().sum()
```

```python
#Tipos de datos dentro de las columnas categoricas
for col in df.select_dtypes(include='O').columns:
    print(df[col].value_counts())
    print('----------------------------')
```

```python
#Convertimos todos los datos en minusculas
sp.valores_minus(df)
```

```python
#Generamos la lista en las cuales queremos cambiar "," por "." y 
coma = ['cons.price.idx','euribor3m','nr.employed','cons.conf.idx']

sp.comas(df,coma)
```

```python
# Convertimos las siguientes columnas a float64 
columnas_float = ['cons.price.idx', 'cons.conf.idx', 'euribor3m', 'nr.employed']
sp.cambiar_tipos(df, columnas_float)
```

```python
# Convertimos las siguientes columnas a booleanos
columnas_booleanas = ['default', 'housing', 'loan']
sp.convertir_a_booleanos(df, columnas_booleanas)
```

```python
#Cambiamos el formato de la columna 'date' a fecha
df['date'] = df['date'].apply(sp.convertir_fecha)
```

```python
#Eliminamos los puntos de la columna job
df['job'] = df['job'].astype(str).str.replace('.', '', regex=False)
```

```python
#Convertimos todos basic. en basic puesto que los estudios hasta esas edades no las considero relevantes como cateogrías diferenciadas 
df["education"] = df["education"].replace({"basic.9y": "basic", "basic.4y": "basic", "basic.6y": "basic"})
```

```python
# Eliminar columnas 'latitude' y 'longitude' puesto que no están ni las intrucciones que nos han dado y parecen irrelevantes
df = df.drop(columns=['latitude', 'longitude'])
```
